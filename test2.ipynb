```python
"""texttools.py"""

def count_consonants(text):
    vowels = "aeiouAEIOU"
    count = 0
    for char in text:
        if char.isalpha() and char not in vowels:
            count += 1
    return count

def reverse_text(text):
    return text[::-1]

text = input("Enter text: ")

print("consonants =", count_consonants(text))
print("Reverse =", reverse_text(text))
```

```python
# main.py
import texttools

# ข้อความตัวอย่างสำหรับทดสอบ
message = "Hello World"

print(f"ข้อความดั้งเดิม: '{message}'")
print("-" * 35

# 1. ทดสอบนับพยัญชนะ
consonant_count = texttools.count_consonants(message)
print(f"จำนวนพยัญชนะ: {consonant_count} ตัว")


# 2. ทดสอบข้อความกลับด้าน
reversed_msg = texttools.reverse_text(message)
print(f"ข้อความกลับด้าน: '{reversed_msg}'")
```

```python
# Given lists
list1 = [1, 2, 3]
list2 = [3, 4, 5]

# Concatenate lists
for i in range(len(list1)):
    if list1[i] == list2[0]:
        concatenated_list = list1 + list2[1:]
        break
    elif list1[i] == list2[-1]:
        concatenated_list = list1 + list2[:-1]
        break
    elif list1[0] == list2[i]:
        concatenated_list = list1[1:] + list2
        break
else:
    concatenated_list = list1 + list2
print("Concatenated list:", concatenated_list)


```
