Question 3-3

```python
import numpy as np

data = np.genfromtxt('data_with_missing.txt', delimiter=',', filling_values=np.nan)

col_means = np.nanmean(data, axis=0)
inds = np.where(np.isnan(data))
data[inds] = np.take(col_means, inds[1])
print(data)
```

```python
import pandas as pd
import numpy as np

data = {
    'EmployeeID': [1, 3, 2, 4, np.nan],
    'Name': ['John Doe', 'Emily Davis', 'Jane, 'Michael, np.nan],
    'DepartmentID': [101, 101, 102, 103, 104],
    'DepartmentName': [Eng, Eng, 'Marketing', 'Sales', 'HR']
}
merged_right = pd.DataFrame(data)

print("Before Handling Missing Data:")
print(merged_right)

missing_data = merged_right[merged_right['EmployeeID'].isna() | merged_right['Name'].isna()]
print("\nRows with Missing Data:")
print(missing_data)
```

```python
import pandas as pd
import numpy as np

data = {
    'EmployeeID': [1, 3, 2, 4, np.nan],
    'Name': ['John Doe', 'Emily Davis', 'Jane Smith', 'Michael Brown', np.nan],
    'DepartmentID': [101, 101, 102, 103, 104],
    'DepartmentName': ['Engineering', 'Engineering', 'Marketing', 'Sales', 'HR']
}
merged_right = pd.DataFrame(data)

print("Before Handling Missing Data:")
print(merged_right)

missing_data = merged_right[merged_right['EmployeeID'].isna() | merged_right['Name'].isna()]
print("\nRows with Missing Data:")
print(missing_data)
```

```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 32, 45, 28],
    'Gender': ['F', 'M', 'M', 'M'],
    'Occupation': ['Engineer', 'Doctor', 'Artist', 'Lawyer']
}
df = pd.DataFrame(data)

selected_columns = df[['Name', 'Occupation']]
print("Selected Columns:")
print(selected_columns)

filtered_rows = df[df['Age'] > 30]
print("\nFiltered Rows (Age > 30):")
print(filtered_rows)
```

```python
import pandas as pd

# Sample DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 32, 45, 28],
    'Gender': ['F', 'M', 'M', 'M'],
    'Occupation': ['Engineer', 'Doctor', 'Artist', 'Lawyer']
}
df = pd.DataFrame(data)

# Group by 'Gender' and calculate the average 'Age'
group_by_gender = df.groupby('Gender')['Age'].mean().reset_index()
print(group_by_gender)
```

```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 32, 45, 28],
    'Gender': ['F', 'M', 'M', 'M'],
    'Occupation': ['Engineer', 'Doctor', 'Artist', 'Lawyer']
}
df = pd.DataFrame(data)
df.set_index('Name', inplace=True)

charlie_row = df.loc['Charlie']
print("Row for Charlie:")
print(charlie_row)

selected_data = df.loc[['Alice', 'Bob'], ['Age', 'Occupation']]
print("\nSelected data for Alice and Bob:")
print(selected_data)
```
