Question 3-2

```python
price = int(input())
paid = int(input())

total_change = paid - price
change = total_change

best_total_coins = float('inf')
best_b100 = 0
best_b50 = 0
best_b20 = 0
best_coin10 = 0
best_coin5 = 0
best_coin1 = 0

for b100 in range((change // 100) + 1):
    rem100 = change - (b100 * 100)
    for b50 in range((rem100 // 50) + 1):
        rem50 = rem100 - (b50 * 50)
        for b20 in range((rem50 // 20) + 1):
            rem20 = rem50 - (b20 * 20)
            
            coin10 = rem20 // 10
            rem_coin = rem20 % 10
            coin5 = rem_coin // 5
            coin1 = rem_coin % 5
            
            total_coins = coin10 + coin5 + coin1
            if total_coins < best_total_coins:
                best_total_coins = total_coins
                best_b100 = b100
                best_b50 = b50
                best_b20 = b20
                best_coin10 = coin10
                best_coin5 = coin5
                best_coin1 = coin1

print(f"total_change: {total_change} baht")
print(f"change 100 cash : {best_b100} bill")
print(f"change 50 cash : {best_b50} bill")
print(f"change 20 cash : {best_b20} bill")
print(f"change 10 coin : {best_coin10} coin")
print(f"change 5 coin : {best_coin5} coin")
print(f"change 1 coin : {best_coin1} coin")
print(f"--> total coins: {best_total_coins} coins")
```
