Question 1

```python
electric_use = int(input("Enter the amount of electricity used in หน่วย: "))
if electric_use >= 0 and electric_use <= 50:
    bill = electric_use * 3
elif electric_use >= 51 and electric_use <= 150:
    bill = electric_use * 4
elif electric_use >= 151 and electric_use <= 400:
    bill = electric_use * 5
elif electric_use > 400:
    bill = electric_use * 6
if electric_use < 0:
    print("Invalid input.")

print(f"The total cost for {electric_use} หน่วย is: {bill} baht.")
```
