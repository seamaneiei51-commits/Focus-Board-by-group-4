Question 3-6

```python
# Student-score dictionary
scores_sem1 = {'Alice': 85, 'Bob': 90, 'Charlie': 78}
scores_sem2 = {'Bob': 88, 'Charlie': 82, 'David': 91}

# Combine dictionary
combined_scores = scores_sem1.copy()
for student, score in scores_sem2.items():
    if student in combined_scores:
        combined_scores[student] += score
    else:
        combined_scores[student] = score

print("Total Score:", combined_scores)
```

Question 3-7

```python
# Given string
text = "hello world"

# Initialize an empty dictionary for character counts
char_count = {}

# Count occurrences of each character
for char in text:
    if char in char_count:
        char_count[char] += 1
    else:
        char_count[char] = 1

print("Character counts:", char_count)
```
