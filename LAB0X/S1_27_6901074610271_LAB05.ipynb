Question 1-8

```python
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

Question 3-2

```python
def count_vowels(text):
    vowels = "aeiouAEIOU"
    count = 0
    for char in text:
        if char.isalpha() and char in vowels:
            count += 1
    return count

text = input("Enter text: ")

print("your text is:", text)
print("veowels =", count_vowels(text))
```
