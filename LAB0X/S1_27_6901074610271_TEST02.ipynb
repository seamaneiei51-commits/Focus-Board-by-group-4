```python
products = {
    "1": ("foods", 50),
    "2": ("cloths", 100),
    "3": ("toys", 200),
    "4": ("electronics", 400)
}

payment_methods = {
    "1": "credit card",
    "2": "prompt pay",
    "3": "cash at delivery",
}

def choose_product():
    print("\nProduct List")
    for key, (name, price) in products.items():
        print(f"{key}. {name} Price {price} Baht")

    while True:
        choice = input("Choose product number: ")
        if choice in products:
            return choice
        print("Invalid input. Please choose a number between 1-4")

def get_quantity():
    while True:
        try:
            quantity = int(input("Enter product quantity: "))
            if quantity > 0:
                return quantity
            print("Quantity must be greater than 0")
        except ValueError:
            print("Please enter an integer")

def choose_payment():
    print("\nPayment Methods")
    for key, name in payment_methods.items():
        print(f"{key}. {name}")

    while True:
        choice = input("Choose payment method: ")
        if choice in payment_methods:
            return choice
        print("Invalid input. Please choose a number between 1-3")

product_id = choose_product()
quantity = get_quantity()
product_name, unit_price = products[product_id]
subtotal = unit_price * quantity

if subtotal > 1000:
    discount = subtotal * 0.10
else:
    discount = 0

payment_id = choose_payment()
shipping_cost = 50 if payment_id == "3" else 0

total = subtotal - discount + shipping_cost
print(f"\nProduct: {product_name}")
print(f"Unit Price: {unit_price} Baht")
print(f"Quantity: {quantity}")
print(f"Payment Method: {payment_methods[payment_id]}")
print(f"Subtotal: {subtotal} Baht")
if discount > 0:
    print(f"Discount: {discount} Baht")
print(f"Shipping Cost: {shipping_cost} Baht")
print(f"Total: {total} Baht")
```
